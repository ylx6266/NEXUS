```python
import numpy as np, pandas as pd
from scipy.spatial.distance import jensenshannon

df=pd.read_csv("table1_2.csv",index_col=0).apply(pd.to_numeric,errors="coerce").dropna(how="all")
M=["I-II(MAE)","I-III(MAE)","II-III(MAE)"]
S=["I-SC","II-SC","III-SC"]
R=["I-II(pearson)","I-III(pearson)","II-III(pearson)"]

geo=lambda r:(1-np.asarray(r,float))/(1-np.asarray(r,float)).sum()
raw=geo(df.loc["raw_functional_data",R])

# Table 1
P=["SC","NO_SC","random_SC","weight_preserved","degree_preserved","uniformed"]
T1=pd.DataFrame(index=P)
T1["Mean_MAE"]=df.loc[P,M].mean(1)
T1["Reduction_vs_NoSC"]=1-T1["Mean_MAE"]/T1.loc["NO_SC","Mean_MAE"]
T1.loc["NO_SC","Reduction_vs_NoSC"]=np.nan
T1["JS_to_raw"]=[jensenshannon(geo(df.loc[x,R]),raw,base=2) for x in P]
T1["Mean_absFC_SC_r"]=df.loc[P,S].mean(1)

print("\nTABLE 1")
print(T1.round({"Mean_MAE":4,"Reduction_vs_NoSC":3,"JS_to_raw":3,"Mean_absFC_SC_r":3}))

# Table 2
rows={
"Raw input, Pearson":df.loc["raw_functional_data",R],
"Pearson FC":df.loc["FC-pearson",R],
"Spearman FC":df.loc["FC-spearman",R],
"Partial-LW FC":df.loc["FC-partial-LW",R],
"Mutual information FC":df.loc["FC-mutualnfo",R],
"NEXUS":df.loc["SC",R]
}

T2=pd.DataFrame([
    [k,*np.asarray(v,float),*geo(v)] for k,v in rows.items()
],columns=["Representation","r_I-II","r_I-III","r_II-III",
           "Sep_I-II","Sep_I-III","Sep_II-III"]).set_index("Representation")

print("\nTABLE S2")
print(T2.round(3))
```
