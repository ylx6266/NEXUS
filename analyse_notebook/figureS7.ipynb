```python
import numpy as np
import matplotlib.pyplot as plt
import os

import numpy as np
import matplotlib.pyplot as plt
import os

def plot_dynamic_graph_results(save_dir="./results"):

    if not os.path.exists(save_dir):
        raise FileNotFoundError(save_dir)

    dyn = np.load(os.path.join(save_dir, "dynamic_full_graphs.npy"))
    T, N, _ = dyn.shape

    mask = ~np.eye(N, dtype=bool)
    dyn2 = dyn[:, mask].reshape(T, -1)

    degree = np.sum(dyn, axis=2)
    mean_degree = degree.mean(axis=1)
    std_degree = degree.std(axis=1)

    avg_strength = np.mean(np.abs(dyn2), axis=1)
    pos_ratio = np.mean(dyn2 > 0, axis=1)
    neg_ratio = np.mean(dyn2 < 0, axis=1)

    plt.rcParams.update({
        "font.size": 13,
        "axes.linewidth": 1.2,
        "xtick.major.width": 1.2,
        "ytick.major.width": 1.2,
        "figure.dpi": 600
    })

    fig, axs = plt.subplots(1, 3, figsize=(18, 5))

    axs[0].plot(mean_degree, lw=2)
    axs[0].set_title("Degree Centrality Over Time")
    axs[0].set_xlabel("Time (Normlized)")
    axs[0].set_ylabel("Avg Degree")

    axs[1].plot(avg_strength, lw=2)
    axs[1].set_title("Mean |Connection Weight|")
    axs[1].set_xlabel("Time (Normlized)")
    axs[1].set_ylabel("|Edge| Mean")

    axs[2].plot(pos_ratio, lw=2, label="Positive")
    axs[2].plot(neg_ratio, lw=2, label="Negative")
    axs[2].set_title("Positive / Negative Edge Ratio")
    axs[2].set_xlabel("Time (Normlized)")
    axs[2].legend(frameon=False)

    for ax in axs:
        ax.spines["top"].set_visible(False)
        ax.spines["right"].set_visible(False)

    plt.tight_layout()

    png = os.path.join(save_dir, "dynamic_graph_analysis.png")
    pdf = os.path.join(save_dir, "dynamic_graph_analysis.pdf")
    plt.savefig(png, bbox_inches="tight")
    plt.savefig(pdf, bbox_inches="tight")
    plt.show()

    return png, pdf


plot_dynamic_graph_results(save_dir="./result/wo_layer/stage1/tuning/bs8_dp0.1_sw1.0")
```

```python
import numpy as np
import matplotlib.pyplot as plt
import os

import numpy as np
import matplotlib.pyplot as plt
import os

def plot_training_epoch_metrics(save_dir="./results"):

    train_path = os.path.join(save_dir, "train_losses.npy")
    val_path = os.path.join(save_dir, "val_losses.npy")
    graph_path = os.path.join(save_dir, "epoch_avg_graphs.npy")

    if not (os.path.exists(train_path) and os.path.exists(val_path) and os.path.exists(graph_path)):
        raise FileNotFoundError(save_dir)

    train_losses = np.load(train_path)
    val_losses = np.load(val_path)
    graphs = np.load(graph_path)

    E, N, _ = graphs.shape
    mask = ~np.eye(N, dtype=bool)
    g = graphs[:, mask].reshape(E, -1)

    eps = 1e-4
    mean_abs = np.mean(np.abs(g), axis=1)
    pos_ratio = np.mean(g > 0, axis=1)
    neg_ratio = np.mean(g < 0, axis=1)

    plt.rcParams.update({
        "font.size": 13,
        "axes.linewidth": 1.2,
        "xtick.major.width": 1.2,
        "ytick.major.width": 1.2,
        "figure.dpi": 600
    })

    fig, axs = plt.subplots(1, 3, figsize=(18, 5))

    axs[0].plot(train_losses, lw=2, label="Train")
    axs[0].plot(val_losses, lw=2, label="Validation")
    axs[0].set_title("Loss vs Epoch")
    axs[0].set_xlabel("Epoch")
    axs[0].set_ylabel("Loss")
    axs[0].legend(frameon=False)

    axs[1].plot(mean_abs, lw=2)
    axs[1].set_title("Mean |Connection| per Epoch")
    axs[1].set_xlabel("Epoch")
    axs[1].set_ylabel("Mean |A|")

    axs[2].plot(pos_ratio, lw=2, label="Positive")
    axs[2].plot(neg_ratio, lw=2, label="Negative")
    axs[2].set_title("Positive / Negative Ratio")
    axs[2].set_xlabel("Epoch")
    axs[2].set_ylabel("Proportion")
    axs[2].legend(frameon=False)

    for ax in axs:
        ax.spines["top"].set_visible(False)
        ax.spines["right"].set_visible(False)

    plt.tight_layout()

    png = os.path.join(save_dir, "training_epoch_metric_plot.png")
    pdf = os.path.join(save_dir, "training_epoch_metric_plot.pdf")
    plt.savefig(png, bbox_inches="tight")
    plt.savefig(pdf, bbox_inches="tight")
    plt.show()

    return png, pdf

###############################
plot_training_epoch_metrics(save_dir="./result/wo_layer/stage1/tuning/bs8_dp0.1_sw1.0")
```
